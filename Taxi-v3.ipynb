# Libraries

```python
import gym
import numpy as np
```

### Initialize environment

```python
env = gym.make('Taxi-v3')
```

#### Hyperparameters

```python
alpha = 0.1            # Learning rate
gamma = 0.6            # Discount factor
epsilon = 1.0          # Exploration rate
max_epsilon = 1.0
min_epsilon = 0.1
epsilon_decay = 0.99
episodes = 1000
max_steps = 100

action_size = env.action_space.n
state_size = env.observation_space.n

q_table = np.zeros([state_size, action_size])
```

### Epsilon Gready Action Selection

```python
def epsilon_gready_action_selection(epsilon, q_table, descrit_state):
    random_val = np.random.random()
    if random_val > epsilon:
        action = np.argmax(q_table[descrit_state, :])
    else:
        action = env.action_space.sample()
    
    return action
```

### Compute Next Q Value

```python
def compute_next_q_value(old_q_value, reward, next_optimal_q_value):
    return old_q_value + alpha * (reward + gamma * next_optimal_q_value - old_q_value)
```

### Reduice Epsilon Value Using Decay_rate

```python
def reduice_epsilon(epsilon, episodes):
    return min_epsilon + (max_epsilon - min_epsilon) * np.exp(-epsilon_decay * episodes)
```

# Agent Training 

```python
rewards = []
log_interval = 1000

for episode in range(episodes):
    state = env.reset()[0]
    done = Falseho
    total_reward = 0

    for step in range(max_steps):
        action = epsilon_gready_action_selection(epsilon, q_table, state)
        next_state, reward, terminated, truncated, _ = env.step(action)
        done =  terminated or truncated

        old_q_value = q_table[state, action]
        next_optimal_q_value = np.max(q_table[next_state]) # max(Q(St+1, a)) and St+1 is next state
        next_q_value = compute_next_q_value(old_q_value, reward, next_optimal_q_value)
        q_table[state, action] = next_q_value

        state = next_state
        total_reward = total_reward + reward
        
        if done:
            break

    epsilon = reduice_epsilon(epsilon, episode)
    rewards.append(total_reward)

    if episode%log_interval ==0:
       print(np.sum(rewards))

env.close()
```

### Test the trained agent

```python
state = env.reset()[0]

for _ in range(20):
    action = np.argmax(q_table[state])
    state, reward, terminated, truncated, _ = env.step(action)

    if terminated or truncated:
        break
        
env.close()
```
